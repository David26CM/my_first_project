```python
# Ejercicio 1
#1. Buscar el mayor y el menor valor
#Dada la tupla numeros = (15, 8, 42, 3, 27, 19, 5), obtén el número mayor, el menor y la
#diferencia entre ambos.

numeros = (15, 8, 42, 3, 27, 19, 5)

mayor = max(numeros)
menor = min(numeros)

diferencia = mayor - menor

print("Número mayor: ", mayor)
print("Número menor: ", menor)
print("Diferencia entre ambos: ", diferencia)
```

```python
# Ejercicio 2
#2. Contar ocurrencias
#Dada la tupla datos = (5, 8, 5, 1, 9, 5, 2, 8, 5, 3), solicita un número al usuario y muestra
#cuántas veces aparece sin convertir la tupla en lista.

datos = (5, 8, 5, 1, 9, 5, 2, 8, 5, 3)

numero = int(input("Ingresa un número: "))

repeticiones = datos.count(numero)

print(f"El número {numero} aparece {repeticiones} veces en la tupla")

```

```python
# Ejercicio 3
#3. Encontrar la primera posición
#Dada la tupla nombres = ('Ana','Luis','Pedro','María','Luis','Carlos'), solicita un nombre y
#muestra su primera posición o un mensaje si no existe.

nombres = ('Ana','Luis','Pedro','María','Luis','Carlos')

buscar_nombre = input("Ingresa tu nombre: ")

if buscar_nombre in nombres:
    primera_posicion = nombres.index(buscar_nombre)
    print(f"La primera posición de '{buscar_nombre}' es el indice: {primera_posicion}")
else:
    print(f"El nombre '{buscar_nombre}' no existe en la tupla")
```

```python
#Ejercicio 4
#4. Estadísticas de una tupla
#Con notas = (18,15,20,11,14,17,19), calcula promedio, máxima, mínima y cantidad de
#aprobados.

notas = (18,15,20,11,14,17,19)

prom = sum(notas) / len (notas)
maxima = max(notas)
minima = min(notas)

aprobados = 0
for nota in notas:
    if nota >= 12:
        aprobados += 1

print(F"Promedio de notas: {(prom)}")
print(f"La nota máxima: {maxima}")
print(f"La nota mínima: {minima}")
print(f"Cantidad de aprobados: {aprobados}")
```

```python
#Ejercicio 5
#5. Tuplas dentro de una lista
#Con ventas=[('Lunes',120),('Martes',150),('Miércoles',95),('Jueves',180),('Viernes',210)],
#calcula el día con mayor venta, el total y el promedio.

ventas=[('Lunes',120),('Martes',150),('Miércoles',95),('Jueves',180),('Viernes',210)]

total_ventas = 0
dia_mayor_venta = ""
monto_mayor_venta = -1

for dia, monto in ventas:
    total_ventas += monto

    if monto > monto_mayor_venta:
        monto_mayor_venta = monto
        dia_mayor_venta = dia

prom_ventas = total_ventas / len(ventas)

print(f"El día con mayor venta: {dia_mayor_venta} ({monto_mayor_venta})")
print(f"Total de ventas: {total_ventas}")
print(f"Promedio de ventas diario: {(prom_ventas)}")
```
