```python
#### Funciones u operadores en strings

- Longitud de mi cadena (len)
- Convertir a mayúsculas (upper)
- Convertir a minúsculas (lower)
- Reemplazar un valor por otro (replace)
- "texto" -> (strip) -> t"texto"TextDisplayObject
- "texto 1 texto 2" -> (split) -> ["texto 1", "texto 2"]
```

```python
nombre = "david"
print("Cuantas letras tiene mi nombre: ", len(nombre))
print(len(nombre))
```

```python
print(nombre.upper())
```

```python
print(nombre.lower())
```

```python
print(
    nombre.replace("d", "D")
)
```

```python
nombre = "David                 "
print(len(nombre))

nombre_limpio = nombre.strip()

print(len(nombre_limpio))

nombre = "               DAVID              "
print(len(nombre))
nombre_limpio = nombre.strip()
print(len(nombre_limpio))
```

```python
nombres = "David Ccallo"

nombres_separados = nombres.split(" ")
print(nombres_separados)

nombres = "DavidxxCcallo"
nombres_separados = nombres.split("xx")
print(nombres_separados)
```
