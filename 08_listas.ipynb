```python
import random

numeros_random =list()
for _ in range(10):
    numeros_random.append(random.randint(1, 20))
    
print(numeros_random)
```

```python
sorted (numeros_random)
```

```python
sorted(numeros_random, reverse=True)
```

```python
numeros_random.sort()
```

```python
numeros_random.sort(reverse=True)
print(numeros_random)
```
