```sql
CREATE TABLE IF NOT EXISTS productos (
    id INT PRIMARY KEY,
    nombre VARCHAR(100),
    precio DECIMAL(10,2),
    stock INT,
    categoria_id INT
);
```

```sql
---EJERCICIO 01
SELECT nombre, precio, stock, categoria_id 
FROM productos 
ORDER BY precio desc;
```

```sql
--EJERCICIO 02
SELECT * 
FROM clientes 
WHERE (pais = 'ES' OR pais = 'MX') 
  AND nombre LIKE 'M%';
```

```sql
--EJERCIO 03
SELECT * 
FROM productos 
WHERE precio >= 20 
  AND precio <= 60;
```

```sql
--EJERCICIO 04
SELECT id, nombre, email, pais, 
       IFNULL(telefono, 'Sin teléfono') AS contacto
FROM clientes 
WHERE telefono IS NULL;
```

```sql
--EJERCICIO 05
SELECT nombre, precio, (precio * 1.21) AS iva_incluido 
FROM productos 
ORDER BY precio DESC 
LIMIT 10;
```

```sql
--EJERCICIO 06
SELECT nombre, precio,
    CASE 
        WHEN precio <= 10 THEN 'Económico'
        WHEN precio <= 300 THEN 'Normal'
        ELSE 'Costoso'
    END AS tipo_precio
FROM productos;
```

```sql
--EJERCICIO 07
SELECT pais 
FROM clientes 
GROUP BY pais;
```
