```python
# EJERCICIO 01
class Persona:
    def __init__(self, nombre: str, edad: int) -> None:
        self.nombre = nombre
        self.edad = edad

    def saludar(self):
        print(f"Hola, soy {self.nombre} y tengo {self.edad} años")


jorge = Persona("Jorge", 10)
maria = Persona("Maria", 20)
luis = Persona("Luis", 15)

jorge.saludar()
maria.saludar()
luis.saludar()
```

```python
# EJERCICIO 02

class Rectangulo:
    def __init__(self, base: float, altura: float) -> None:
        self.base = base
        self.altura = altura

    def area(self) -> float:
        return self.base * self.altura

    def perimetro(self) -> float:
        return 2 * (self.base + self.altura)

    def __str__(self) -> str:
        return f"Rectangulo de base: {self.base} y altura: {self.altura}"

mi_rectangulo = Rectangulo(3, 5)
print(mi_rectangulo.area())
print(mi_rectangulo.perimetro())
print(mi_rectangulo)
```

```python
# EJERCICIO 05

class Libro:
    def __init__(self, titulo: str, autor: str, precio: float) -> None:
        self.titulo = titulo
        self.autor = autor
        self.__precio = precio

    def get_precio(self) -> float:
        return self.__precio

    def set_precio(self, nuevo_precio: float) -> None:
        if nuevo_precio > 0:
            self.__precio = nuevo_precio
        else:
            print(f"Error el nuevo precio es 0 o negativo: {nuevo_precio}")


mi_libro = Libro(titulo="Mi planta de naranja Lima", autor="Jose Mauro de Vasconcelos", precio=20)
# Libro.__init__(titulo="Mi planta de naranja Lima", autor="Jose Mauro de Vasconcelos", precio=20)
# print(mi_libro.__precio)

print(mi_libro.get_precio())

mi_libro.set_precio(50)

print(mi_libro.get_precio())


mi_libro.set_precio(-100)
```

```python
# EJERCICIO 06
class Animal:
    def __init__(self, nombre: str, edad: int) -> None:
        self.nombre = nombre
        self.edad = edad

    def hacer_sonido(self):
        print("Grrrrr")


class Perro(Animal):
    def __init__(self, nombre: str, edad: int) -> None:
        super().__init__(nombre, edad)

    def hacer_sonido(self):
        print("Gua gua!")

class Gato(Animal):
    def __init__(self, nombre: str, edad: int) -> None:
        super().__init__(nombre, edad)

    def hacer_sonido(self):
        print("Miau Miau!")


perro_1 = Perro("Hachiko", 3)
perro_2 = Perro("Fido", 5)
gato_1 = Gato("Lucifer", 10)
vaca_1 = Animal("Lola", 14)

animales = [perro_1, perro_2, gato_1, vaca_1]

for animal in animales:
    animal.hacer_sonido()
```

```python
# EJERCICIO 08

class Reproductor:
    def __init__(self):
        self.playlist =[]
        self.posicion =0
    def agregar_cancion(self,cancion:str):
        self.playlist.append(cancion)

    def cancion_actual(self)-> str:
        if not self.playlist:
            return "No hay canciones en la playlits"
        return self.playlist[self.posicion]
    
    def siguiente(self) -> str:
        self.posicion +=1
        if self.posicion<len(self.playlist):
            cancion_actual=self.cancion_actual()
        else:
            
            self.posicion -=1
            cancion_actual=self.cancion_actual()
       
        return cancion_actual
    
    def anterior(self) -> str:
        self.posicion -=1
        if self.posicion >=0:
            cancion_actual=self.cancion_actual()
        else:
            self.posicion += 1
            self.cancion_actual=self.cancion_actual()

        return self.cancion_actual
mi_reproductor= Reproductor()
mi_reproductor.agregar_cancion("is war")
mi_reproductor.agregar_cancion("lfav")
mi_reproductor.siguiente()
mi_reproductor.cancion_actual()
mi_reproductor.siguiente()
mi_reproductor.anterior()
mi_reproductor.cancion_actual()
mi_reproductor.siguiente()
```

```python
# EJERCICIO 09:


class Motor:
    def __init__(self, tipo: str, caballo: float)-> None:
        self.tipo=tipo
        self.caballo=caballo
        self.encendido=False
    def arrancar(self):
        if self.encendido:
            return "El motor ya esta encendido"
        
        self.encendido=True
        return f"El motor  {self.tipo} de {self.caballo}HP se encendio"
    
class Vehiculo:
    def __init__(self, marca: str, motor: Motor)->None:
        self.marca=marca
        self.motor=motor
    def arrancar(self)->str:
        return self.motor.arrancar()
motor= Motor("gasolina", 150)
vehiculo= Vehiculo("ford",motor)

print(vehiculo.arrancar())
print(vehiculo.arrancar())
```
