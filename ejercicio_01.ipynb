## Calculadora de propinas

```python
##Calculadora ejemplo 01
total_compra = input("Ingresa el total de la compra:")
total = float(total_compra)
propina = total * 0.05
Total_cuenta = total + propina

print("El total de la compra es: ", total)
print("La propina es: ", propina)
print(" El total de la cuenta es: ", Total_cuenta)
```
