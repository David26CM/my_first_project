## & para la intersección (lo que tienen en común).
## - para la diferencia (lo que tiene uno que no tiene el otro).
## | para la unión (juntar todo sin repetir).

```python
# Ejercicio 1
#Encuentra:
#- Los estudiantes inscritos en ambos cursos.
#- Los estudiantes que solo están en Python.
#- Los estudiantes que solo están en JavaScript.
#- Todos los estudiantes sin repetir nombres.
estudiantes_python = {
"Ana", "Luis", "María", "Carlos", "Sofía", "Diego"
}
estudiantes_javascript = {
"María", "Carlos", "Pedro", "Lucía", "Diego", "Valeria"
}
```

```python
print(" Los estudiantes inscritos en ambos cursos: ", estudiantes_python & estudiantes_javascript)
print(" Los estudiantes que solo están en Python: ", estudiantes_python - estudiantes_javascript)
print(" Los estudiantes que solo están en JavaScript: ", estudiantes_javascript - estudiantes_python)
print(" Todos los estudiantes sin repetir nombres: ", estudiantes_python | estudiantes_javascript)
```

```python
# Ejercicio 2
#Determina:
#- Los productos disponibles en ambos almacenes.
#- Los productos exclusivos del almacén de Arequipa.
#- Los productos exclusivos del almacén de Lima.
#- Todos los productos disponibles sin repetir.
#- Si ambos almacenes tienen exactamente el mismo inventario.
# == sirve para preguntar: "¿Estos dos conjuntos son exactamente iguales?"

almacen_arequipa = {
"laptop", "teclado", "mouse", "monitor", "impresora", "audifonos"
}
almacen_lima = {
"laptop", "mouse", "monitor", "tablet", "webcam", "parlantes"
}
```

```python
print("Los productos diponibles en ambos almacenes: ", almacen_arequipa & almacen_lima)
print("Los productos exclusivos del almacén de Arequipa: ", almacen_arequipa - almacen_lima)
print("Los productos exclusivos del almacén de Lima: ", almacen_lima - almacen_arequipa)
print("Todos los prodcutos disponibles sin repetir: ", almacen_arequipa | almacen_lima)
print("¿Tienen el mismo inventario?: ", almacen_arequipa == almacen_lima)
```

```python
# Ejercicio 3
#Crea un programa que muestre:
#- Si el usuario posee todos los permisos requeridos.
#- Los permisos que le faltan.
#- Los permisos adicionales que tiene.
#- La cantidad total de permisos diferentes.
# issubset() Sirve para preguntar: "¿Todos los elementos de mi conjunto están dentro de este otro conjunto?"

permisos_usuario = {
"leer", "editar", "comentar", "descargar"
}
permisos_requeridos = {
"leer", "editar", "eliminar", "publicar"
}

```

```python
print("¿El usuario tiene todos los permisos requeridos?: ", permisos_requeridos.issubset(permisos_usuario))
print("¿Qué permisos le faltan?: ", permisos_requeridos - permisos_usuario)
print("¿Qué permisos adicionales tiene?: ", permisos_usuario - permisos_requeridos)
print("Cantidad total de permisos diferentes: ", len(permisos_usuario - permisos_requeridos))
```
