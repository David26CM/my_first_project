```python
# Ejercicio 2
numeros = [12, 7, 9, 24, 31, 18, 5, 42, 13, 8]
num_pares = []
for numero in numeros:
    if numero % 2 == 0:
        num_pares.append(numero)
print(f"Los números pares son: {num_pares}")

num_impares = []
for numero in numeros:
    if numero % 2 != 0:
        num_impares.append(numero)
print(f"Los números impares son: {num_impares}")
```

```python
#Ejercicio 5
notas = [14, 18, 11, 16, 9, 13, 20]
def calcular_promedio(notas):
    if len(notas) == 0:
        return 0
    return sum(notas) / len(notas)

def resultado_final(promedio):
    if promedio >= 11:
        return "Aprobó"
    else:
        return "Desaprobó"

print(f"El promedio de las notas es: {calcular_promedio(notas)}")
print(f"El resultado final es: {resultado_final(calcular_promedio(notas))}")
```

```python
#Ejercicio 6 (Filtrar números positivos)
#Dada una lista con números positivos y negativos, crea una nueva lista solo con los números positivos.
numeros = [-5, 8, -2, 15, 0, -9, 23, -1, 7]
numeros_positivos = [n for n in numeros if n > 0]
print(numeros_positivos)
```

```python
#Ejercicio 8: Buscar elementos comunes
#Dadas dos listas, encuentra los elementos que aparecen en ambas.
lista1 = [1, 3, 5, 7, 9, 11]
lista2 = [2, 3, 6, 7, 10, 11]

#Declara lista vacía

elementos_comunes = []

for item_1 in lista1:
    if item_1 in lista2:
        elementos_comunes.append(item_1)

print(f"Los elementos comunes son: {elementos_comunes}")
```

```python
#Ejercicio 9: Buscar elementos únicos entre dos listas
#Dadas dos listas, crea una nueva lista con los elementos que no se repiten entre ambas listas.
lista1 = [4, 8, 12, 16, 20]
lista2 = [8, 10, 12, 18, 20]

elementos_no_comunes = []

for item_1 in lista1:
    if item_1 not in lista2:
        elementos_no_comunes.append(item_1)

for item_2 in lista2:
    if item_2 not in lista1:
        elementos_no_comunes.append(item_2)

print(f"Los elementos no comunes son: {elementos_no_comunes}")
```

```python
#Ejercicio 10: Invertir una lista sin usar reverse()
#Dada una lista de colores, invierte la lista sin usar el método .reverse().
colores = ['rojo', 'azul', 'verde', 'amarillo', 'morado']

ultimo_indice = len(colores) - 1

colores_invertidos = []
while ultimo_indice >= 0:
    colores_invertidos.append(colores[ultimo_indice])
    ultimo_indice = ultimo_indice - 1

print(f"Lista invertidad: {colores_invertidos}")
```

```python
#Ejercicio 13: Contar frecuencia de elementos
#Dada una lista de nombres repetidos, muestra cuántas veces aparece cada nombre.
nombres = ['Ana', 'Luis', 'Ana', 'Carlos', 'Luis', 'Ana', 'María', 'Carlos']
nombres_unicos = []
for nombre in nombres:
    if nombre not in nombres_unicos:
        nombres_unicos.append(nombre)

print(f"Nombres unicos: {nombres_unicos}")

for nombre_unico in nombres_unicos:
    print(
        f"nombre_unico: {nombre_unico} count: {nombres.count(nombre_unico)}"
    )
```

```python
#Ejercicio 16: Dividir una lista en dos partes
#Dada una lista de números, divídela en dos mitades.
numeros = [10, 20, 30, 40, 50, 60, 70, 80]
half_list = []
last_list = []
longitud_lista = len(numeros)
indice_mitad = longitud_lista // 2

for idx, numero in enumerate(numeros):
    if idx < indice_mitad:
        half_list.append(numero)
    else:
        last_list.append(numero)

print(f"Primera mitad: {half_list}")
print(f"Ultima mitad: {last_list}")
```

```python
#usando slicing
print(f"Primera mitad: {numeros[:indice_mitad]}")
print(f"Primera mitad: {numeros[indice_mitad :]}")
```

```python
#Ejercicio 18: Eliminar números menores a 10
#Dada una lista de números, crea una nueva lista eliminando todos los números menores que 10.
numeros = [3, 15, 8, 22, 1, 10, 19, 6, 30]
numeros_gt_10 = [
    numero
    for numero in numeros
    if numero >= 10
]
print(f"Imprimir numeros mayores o iguales a 10: {numeros_gt_10}")
```

```python
# Ejercicio 19: Combinar dos listas alternando elementos
#Dadas dos listas del mismo tamaño, crea una nueva lista alternando los elementos de ambas listas.
numeros = [1, 2, 3, 4]
letras = ['a', 'b', 'c', 'd']
resultado_esperado = [1, 'a', 2, 'b', 3, 'c', 4, 'd']

lista_alternada = []
for numero, letra in zip(numeros, letras):
    lista_alternada.extend([numero, letra])

print(f"Lista alternada: {lista_alternada}")
```

```python
#Ejercicio 20: Crear una lista de estudiantes aprobados
#Dada una lista de estudiantes y otra lista con sus notas, crea una nueva lista solo con los nombres de los
#estudiantes que aprobaron. Considera que la nota mínima aprobatoria es 11.
estudiantes = ['Ana', 'Luis', 'Carlos', 'María', 'Pedro']
notas = [18, 10, 14, 8, 16]

lista_de_aprobados = []
for estudiante, nota in zip(estudiantes, notas):
    if nota > 11:
        lista_de_aprobados.append(estudiante)

print(f"Lista de aprobados: {lista_de_aprobados}")

```
