## Tipos de variables

```python
# Variable texto, cadena o string
nombre: str = "David"
apellido = "Ccallo"

print(nombre + ' '+ apellido)

# variable numericas enteros o int
edad: int = 30

# variable numericas flotantes o float
estatura: float = 1.75

print("Mi edad dentro de 5 años será: ", edad + 5)
print("Mi edad hace 5 años fue: ", edad - 5)
print("El doble de mi edad es: ", edad * 2)
print("La mitad de mi edad es: ", edad / 2)
print("La mitad de mi edad entera es: ", edad // 2)
print("El resto de mi edad es: ", edad % 2)

```
