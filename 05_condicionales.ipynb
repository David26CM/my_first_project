### Condicionales

Si se cumple esta CONDICION:
    - Proceso 1
Si se cumple esta otra CONDICION:
    - Proceso 2
Si no se cumple ninguna de las otras 2 condiciones:
    - Proceso por defecto o por default

```python
edad = 20

if edad >= 18:
    print("soy un adulto")
elif edad <= 13:
    print("soy un adolescente")
else:
    print("soy un niño")
```

```python
temp_sensor_1 = 30
alerta_mayor_15 = 15

if temp_sensor_1 > alerta_mayor_15:
    print("Peligro!")
elif temp_sensor_1 > 10 and temp_sensor_1 < 20:
    print("condicion normal")
```
